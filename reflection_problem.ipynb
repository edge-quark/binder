# Understanding Reflection in a Plane Along a Line

## Problem 5.12

**Given:** Basis vectors $\mathbf{e}_1, \mathbf{e}_2, \mathbf{e}_3$ for $\mathbb{R}^3$

**Find:** Matrix of reflection in the plane $2x_1 - x_2 - 3x_3 = 0$ along the line $\mathbf{x} = t(1, -2, 1)$

---

## 🤔 What Does This Mean Intuitively?

This is NOT a standard reflection! Let me break it down:

### Standard Reflection vs. This Problem

- **Standard reflection in a plane:** You reflect perpendicular to the plane (like a mirror)
- **Reflection in a plane ALONG a line:** You reflect parallel to a specific direction (the line)

### The Intuition

Think of it like this:

1. **The plane** $2x_1 - x_2 - 3x_3 = 0$ is like a "mirror plane"
2. **The line** $\mathbf{x} = t(1, -2, 1)$ tells you the **direction** in which to reflect
3. For any point **not on the plane**, you move along the direction $(1, -2, 1)$ until you hit the plane, then continue the same distance on the other side

### Key Insight

This transformation:
- **Keeps the plane fixed** (points on the plane don't move)
- **Reflects along the line direction** (like sliding along rails parallel to the line)

## 📐 Visual Setup

Let's visualize the geometry:

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.patches import FancyArrowPatch
from mpl_toolkits.mplot3d.proj3d import proj_transform

# Define the plane: 2x₁ - x₂ - 3x₃ = 0
# Normal vector to the plane
n = np.array([2, -1, -3])

# Direction of reflection
d = np.array([1, -2, 1])

print("Plane normal vector n =", n)
print("Reflection direction d =", d)
print("\nDot product n·d =", np.dot(n, d))
```

```python
# Create a 3D plot
fig = plt.figure(figsize=(14, 6))

# First subplot: Show the plane and direction
ax1 = fig.add_subplot(121, projection='3d')

# Create plane mesh
x1 = np.linspace(-3, 3, 20)
x2 = np.linspace(-3, 3, 20)
X1, X2 = np.meshgrid(x1, x2)
# From 2x₁ - x₂ - 3x₃ = 0, we get x₃ = (2x₁ - x₂)/3
X3 = (2*X1 - X2) / 3

# Plot the plane
ax1.plot_surface(X1, X2, X3, alpha=0.3, color='cyan', edgecolor='none')

# Plot the normal vector
ax1.quiver(0, 0, 0, n[0], n[1], n[2], color='red', arrow_length_ratio=0.2, linewidth=2, label='Normal n')

# Plot the reflection direction
ax1.quiver(0, 0, 0, d[0]*2, d[1]*2, d[2]*2, color='green', arrow_length_ratio=0.1, linewidth=2, label='Direction d')

# Show several parallel lines in direction d
t_vals = np.linspace(-2, 2, 20)
for offset in [np.array([1, 0, 0.67]), np.array([-1, 0, -0.67]), np.array([0, 1, -0.33])]:
    line_points = offset + t_vals[:, None] * d
    ax1.plot(line_points[:, 0], line_points[:, 1], line_points[:, 2], 
             'g--', alpha=0.4, linewidth=1)

ax1.set_xlabel('$x_1$')
ax1.set_ylabel('$x_2$')
ax1.set_zlabel('$x_3$')
ax1.set_title('Plane and Reflection Direction')
ax1.legend()

# Second subplot: Show how a point reflects
ax2 = fig.add_subplot(122, projection='3d')

# Plot the plane again
ax2.plot_surface(X1, X2, X3, alpha=0.3, color='cyan', edgecolor='none')

# Choose a point not on the plane
P = np.array([2, 1, 0])

# To find reflection:
# 1. Find where the line P + td intersects the plane
# Plane equation: n·x = 0, so n·(P + td) = 0
# n·P + t(n·d) = 0
# t = -n·P / (n·d)

t_intersect = -np.dot(n, P) / np.dot(n, d)
Q = P + t_intersect * d  # Point on plane

# Reflection is same distance on other side
P_reflected = Q + t_intersect * d  # = P + 2*t_intersect*d

# Plot original point
ax2.scatter(*P, color='blue', s=100, label='Original P', zorder=5)

# Plot intersection point
ax2.scatter(*Q, color='orange', s=100, label='On plane Q', zorder=5)

# Plot reflected point
ax2.scatter(*P_reflected, color='purple', s=100, label="Reflected P'", zorder=5)

# Draw the reflection line
line_points = np.array([P - d, P_reflected + d])
ax2.plot(line_points[:, 0], line_points[:, 1], line_points[:, 2], 
         'g--', linewidth=2, alpha=0.7, label='Reflection path')

# Draw lines from points to plane intersection
ax2.plot([P[0], Q[0]], [P[1], Q[1]], [P[2], Q[2]], 'b-', linewidth=2, alpha=0.7)
ax2.plot([Q[0], P_reflected[0]], [Q[1], P_reflected[1]], [Q[2], P_reflected[2]], 
         'purple', linewidth=2, alpha=0.7)

ax2.set_xlabel('$x_1$')
ax2.set_ylabel('$x_2$')
ax2.set_zlabel('$x_3$')
ax2.set_title('How a Point Reflects')
ax2.legend()

plt.tight_layout()
plt.show()

print(f"\nOriginal point P = {P}")
print(f"Distance to plane along d: t = {t_intersect:.3f}")
print(f"Intersection point Q = {Q}")
print(f"Reflected point P' = {P_reflected}")
```

## 🧮 Mathematical Approach

### Step 1: Understand the Transformation

For a point $\mathbf{P}$, we need to:
1. Find where the line $\mathbf{P} + t\mathbf{d}$ intersects the plane
2. Continue the same distance on the other side

### Step 2: Derive the Formula

The plane equation is $\mathbf{n} \cdot \mathbf{x} = 0$ where $\mathbf{n} = (2, -1, -3)$

To find the intersection:
$$\mathbf{n} \cdot (\mathbf{P} + t\mathbf{d}) = 0$$
$$\mathbf{n} \cdot \mathbf{P} + t(\mathbf{n} \cdot \mathbf{d}) = 0$$
$$t = -\frac{\mathbf{n} \cdot \mathbf{P}}{\mathbf{n} \cdot \mathbf{d}}$$

The reflected point is:
$$\mathbf{P}' = \mathbf{P} + 2t\mathbf{d} = \mathbf{P} - 2\frac{\mathbf{n} \cdot \mathbf{P}}{\mathbf{n} \cdot \mathbf{d}}\mathbf{d}$$

### Step 3: Matrix Form

We can write:
$$\mathbf{P}' = \mathbf{P} - 2\frac{(\mathbf{n}^T \mathbf{P})(\mathbf{d})}{\mathbf{n} \cdot \mathbf{d}}$$

This is equivalent to:
$$\mathbf{P}' = \left(\mathbf{I} - 2\frac{\mathbf{d}\mathbf{n}^T}{\mathbf{n} \cdot \mathbf{d}}\right)\mathbf{P}$$

So the reflection matrix is:
$$R = \mathbf{I} - 2\frac{\mathbf{d}\mathbf{n}^T}{\mathbf{n} \cdot \mathbf{d}}$$

```python
# Calculate the reflection matrix
n = np.array([2, -1, -3])
d = np.array([1, -2, 1])

# Reflection matrix formula: R = I - 2(d⊗n)/(n·d)
# where d⊗n is the outer product
n_dot_d = np.dot(n, d)

print(f"n·d = {n_dot_d}")
print("\nOuter product d⊗n:")
outer_product = np.outer(d, n)
print(outer_product)

# Reflection matrix
R = np.eye(3) - 2 * outer_product / n_dot_d

print("\n" + "="*50)
print("REFLECTION MATRIX R:")
print("="*50)
print(R)
print("\nAs fractions:")
for i in range(3):
    for j in range(3):
        print(f"R[{i},{j}] = {R[i,j]:.4f} = {int(R[i,j]*7)}/7", end="  ")
    print()
```

## ✅ Verification

Let's verify our matrix has the right properties:

```python
print("VERIFICATION TESTS")
print("="*50)

# Test 1: Points on the plane should be fixed
print("\n1. Points on the plane should stay fixed:")
# A point on the plane: 2x₁ - x₂ - 3x₃ = 0
# Let's use x₁=1, x₂=2, then x₃=0
point_on_plane = np.array([1, 2, 0])
print(f"   Point on plane: {point_on_plane}")
print(f"   Check: 2({point_on_plane[0]}) - {point_on_plane[1]} - 3({point_on_plane[2]}) = {2*point_on_plane[0] - point_on_plane[1] - 3*point_on_plane[2]}")
reflected = R @ point_on_plane
print(f"   After reflection: {reflected}")
print(f"   Fixed? {np.allclose(point_on_plane, reflected)}")

# Test 2: The direction d should be fixed (eigenvector with eigenvalue 1)
print("\n2. Direction vector d should be fixed:")
print(f"   d = {d}")
Rd = R @ d
print(f"   R·d = {Rd}")
print(f"   Fixed? {np.allclose(d, Rd)}")

# Test 3: Apply twice should give original (R² = I)
print("\n3. Applying reflection twice should give identity:")
R2 = R @ R
print(f"   R² = I? {np.allclose(R2, np.eye(3))}")
print(f"   Max difference from identity: {np.max(np.abs(R2 - np.eye(3)))}")

# Test 4: Normal direction behavior
print("\n4. The normal vector n has special behavior:")
print(f"   n = {n}")
Rn = R @ n
print(f"   R·n = {Rn}")
# Check if Rn is parallel to n
ratio = Rn / n
print(f"   Ratio R·n/n = {ratio}")
print(f"   (Should be constant - eigenvalue)")
```

## 🎯 Key Insights

### What Makes This Problem Special?

1. **Two fixed structures:**
   - The plane itself (all points on it stay fixed)
   - The direction $\mathbf{d}$ (parallel lines stay parallel)

2. **The formula:** $R = I - 2\frac{\mathbf{d}\mathbf{n}^T}{\mathbf{n} \cdot \mathbf{d}}$
   - This is an **oblique reflection** (not perpendicular)
   - It's a **linear transformation** (matrix multiplication)

3. **Eigenstructure:**
   - Eigenvalue $\lambda = 1$ with multiplicity 2 (the plane)
   - One more eigenvalue (related to reflection)

### Comparison with Standard Reflection

| Standard Reflection | This Problem |
|---------------------|---------------|
| Perpendicular to plane | Along specific direction |
| Formula: $I - 2\frac{\mathbf{n}\mathbf{n}^T}{\|\mathbf{n}\|^2}$ | Formula: $I - 2\frac{\mathbf{d}\mathbf{n}^T}{\mathbf{n} \cdot \mathbf{d}}$ |
| $\mathbf{d} = \mathbf{n}$ | $\mathbf{d} \neq \mathbf{n}$ (oblique) |

## 🎨 Interactive Visualization

Let's test the reflection on various points:

```python
# Test multiple points
test_points = np.array([
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1],
    [2, 1, 0],
    [1, 1, 1],
])

fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Plot the plane
x1 = np.linspace(-2, 3, 20)
x2 = np.linspace(-2, 3, 20)
X1, X2 = np.meshgrid(x1, x2)
X3 = (2*X1 - X2) / 3
ax.plot_surface(X1, X2, X3, alpha=0.2, color='cyan', edgecolor='none')

# Plot normal and direction vectors
ax.quiver(0, 0, 0, n[0]*0.5, n[1]*0.5, n[2]*0.5, 
          color='red', arrow_length_ratio=0.2, linewidth=2, label='Normal n')
ax.quiver(0, 0, 0, d[0], d[1], d[2], 
          color='green', arrow_length_ratio=0.2, linewidth=2, label='Direction d')

# For each test point
colors = plt.cm.rainbow(np.linspace(0, 1, len(test_points)))

for i, P in enumerate(test_points):
    P_reflected = R @ P
    
    # Plot original point
    ax.scatter(*P, color=colors[i], s=100, marker='o', edgecolors='black', linewidths=2)
    
    # Plot reflected point
    ax.scatter(*P_reflected, color=colors[i], s=100, marker='s', edgecolors='black', linewidths=2)
    
    # Draw reflection path
    ax.plot([P[0], P_reflected[0]], [P[1], P_reflected[1]], [P[2], P_reflected[2]], 
            color=colors[i], linestyle='--', linewidth=2, alpha=0.6)
    
    # Find intersection with plane
    t_int = -np.dot(n, P) / np.dot(n, d)
    Q = P + t_int * d
    ax.scatter(*Q, color=colors[i], s=50, marker='x', linewidths=2)

ax.set_xlabel('$x_1$', fontsize=12)
ax.set_ylabel('$x_2$', fontsize=12)
ax.set_zlabel('$x_3$', fontsize=12)
ax.set_title('Multiple Point Reflections\n(circles = original, squares = reflected, x = on plane)', fontsize=14)
ax.legend(fontsize=10)

plt.tight_layout()
plt.show()

# Print numerical results
print("\nReflection Results:")
print("="*70)
for i, P in enumerate(test_points):
    P_reflected = R @ P
    print(f"Point {i+1}: {P} → {P_reflected}")
    print(f"         Distance moved: {np.linalg.norm(P_reflected - P):.3f}")
    print()
```

## 🎓 Summary: The Intuition

Think of this transformation as:

1. **Imagine rails** running parallel to the direction $\mathbf{d} = (1, -2, 1)$
2. **The plane** acts as a mirror, but you can only slide along these rails
3. **To reflect a point:**
   - Slide it along the rail until you hit the plane
   - Keep sliding the same distance on the other side
4. **Points on the plane** don't move (they're already "at the mirror")
5. **The direction** $\mathbf{d}$ is preserved (rails stay parallel)

This is called an **oblique reflection** or **shear-like reflection** because you're not reflecting perpendicular to the plane, but at an angle determined by $\mathbf{d}$.
