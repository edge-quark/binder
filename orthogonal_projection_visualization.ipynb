# Orthogonal Projection onto a Plane

## Problem Statement
Find the matrix of orthogonal projection on the plane $x_1 + 2x_2 - 2x_3 = 0$

## Mathematical Theory

### Key Concepts
- A plane through the origin can be described by its **normal vector** $\mathbf{n}$
- For the plane $x_1 + 2x_2 - 2x_3 = 0$, the normal vector is $\mathbf{n} = \begin{bmatrix} 1 \\ 2 \\ -2 \end{bmatrix}$
- The **orthogonal projection** of a vector $\mathbf{v}$ onto the plane removes its component perpendicular to the plane

### Derivation

The projection of any vector $\mathbf{v}$ onto the plane can be found by:

1. **Component perpendicular to plane**: The projection of $\mathbf{v}$ onto the normal $\mathbf{n}$ is:
   $$\text{proj}_{\mathbf{n}}(\mathbf{v}) = \frac{\mathbf{n} \cdot \mathbf{v}}{\mathbf{n} \cdot \mathbf{n}} \mathbf{n} = \frac{\mathbf{n}\mathbf{n}^T}{\mathbf{n}^T\mathbf{n}} \mathbf{v}$$

2. **Component parallel to plane**: Subtract the perpendicular component:
   $$\text{proj}_{\text{plane}}(\mathbf{v}) = \mathbf{v} - \text{proj}_{\mathbf{n}}(\mathbf{v}) = \left(I - \frac{\mathbf{n}\mathbf{n}^T}{\mathbf{n}^T\mathbf{n}}\right) \mathbf{v}$$

3. **Projection Matrix**:
   $$P = I - \frac{\mathbf{n}\mathbf{n}^T}{\mathbf{n}^T\mathbf{n}}$$

Let's compute this!

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.patches import FancyArrowPatch
from mpl_toolkits.mplot3d.proj3d import proj_transform

# Set up nice plotting defaults
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 11
```

## Step-by-Step Computation

```python
# Step 1: Define the normal vector from the plane equation x1 + 2x2 - 2x3 = 0
n = np.array([[1], [2], [-2]])
print("Normal vector n:")
print(n)
print()

# Step 2: Compute n·n^T (outer product)
n_nT = n @ n.T
print("n·n^T (outer product):")
print(n_nT)
print()

# Step 3: Compute n^T·n (dot product = magnitude squared)
nT_n = n.T @ n
print("n^T·n (magnitude squared):")
print(nT_n)
print(f"This equals: 1² + 2² + (-2)² = {1**2 + 2**2 + (-2)**2}")
print()

# Step 4: Compute the projection onto the normal
proj_normal = n_nT / nT_n[0, 0]
print("Projection matrix onto normal (n·n^T)/(n^T·n):")
print(proj_normal)
print()

# Step 5: Compute the projection onto the plane
I = np.eye(3)
P = I - proj_normal

print("="*60)
print("FINAL PROJECTION MATRIX onto the plane:")
print("="*60)
print(P)
print()

# Display as fractions for clarity
print("As fractions (with denominator 9):")
print(P * 9)
```

## Verification

Let's verify our projection matrix has the correct properties:
1. **Idempotent**: $P^2 = P$ (applying the projection twice gives the same result)
2. **Symmetric**: $P^T = P$ (orthogonal projections are symmetric)
3. **Projects normal to zero**: $P\mathbf{n} = \mathbf{0}$ (normal vector is perpendicular to plane)

```python
# Property 1: P² = P (idempotent)
P_squared = P @ P
print("Property 1: Is P² = P? (Idempotent)")
print(f"Max difference: {np.max(np.abs(P_squared - P)):.2e}")
print(f"✓ Verified!" if np.allclose(P_squared, P) else "✗ Failed")
print()

# Property 2: P^T = P (symmetric)
print("Property 2: Is P^T = P? (Symmetric)")
print(f"Max difference: {np.max(np.abs(P.T - P)):.2e}")
print(f"✓ Verified!" if np.allclose(P.T, P) else "✗ Failed")
print()

# Property 3: Pn = 0 (projects normal vector to zero)
Pn = P @ n
print("Property 3: Does P·n = 0? (Normal projects to zero)")
print(f"P·n = {Pn.T}")
print(f"Max absolute value: {np.max(np.abs(Pn)):.2e}")
print(f"✓ Verified!" if np.allclose(Pn, 0) else "✗ Failed")
```

## 3D Visualization

Now let's visualize what this projection actually does!

```python
# Helper class for drawing 3D arrows
class Arrow3D(FancyArrowPatch):
    def __init__(self, x, y, z, dx, dy, dz, *args, **kwargs):
        super().__init__((0, 0), (0, 0), *args, **kwargs)
        self._xyz = (x, y, z)
        self._dxdydz = (dx, dy, dz)

    def draw(self, renderer):
        x1, y1, z1 = self._xyz
        dx, dy, dz = self._dxdydz
        x2, y2, z2 = (x1 + dx, y1 + dy, z1 + dz)

        xs, ys, zs = proj_transform((x1, x2), (y1, y2), (z1, z2), self.axes.M)
        self.set_positions((xs[0], ys[0]), (xs[1], ys[1]))
        super().draw(renderer)
        
    def do_3d_projection(self, renderer=None):
        x1, y1, z1 = self._xyz
        dx, dy, dz = self._dxdydz
        x2, y2, z2 = (x1 + dx, y1 + dy, z1 + dz)

        xs, ys, zs = proj_transform((x1, x2), (y1, y2), (z1, z2), self.axes.M)
        self.set_positions((xs[0], ys[0]), (xs[1], ys[1]))
        
        return np.min(zs)
```

```python
# Create figure with the plane
fig = plt.figure(figsize=(14, 10))
ax = fig.add_subplot(111, projection='3d')

# Create the plane: x1 + 2x2 - 2x3 = 0 => x3 = (x1 + 2x2)/2
x1_range = np.linspace(-3, 3, 20)
x2_range = np.linspace(-3, 3, 20)
X1, X2 = np.meshgrid(x1_range, x2_range)
X3 = (X1 + 2*X2) / 2  # From the plane equation

# Plot the plane
ax.plot_surface(X1, X2, X3, alpha=0.3, color='lightblue', edgecolor='none')

# Plot the normal vector
n_flat = n.flatten()
arrow_n = Arrow3D(0, 0, 0, n_flat[0]*1.5, n_flat[1]*1.5, n_flat[2]*1.5,
                  mutation_scale=20, lw=3, arrowstyle="->", color="red")
ax.add_artist(arrow_n)

# Example vectors and their projections
example_vectors = [
    np.array([[2], [1], [0]]),
    np.array([[1], [1], [2]]),
    np.array([[0], [2], [1]]),
]

colors = ['green', 'purple', 'orange']

for i, v in enumerate(example_vectors):
    # Original vector
    v_flat = v.flatten()
    arrow_v = Arrow3D(0, 0, 0, v_flat[0], v_flat[1], v_flat[2],
                      mutation_scale=15, lw=2.5, arrowstyle="->", color=colors[i])
    ax.add_artist(arrow_v)
    
    # Projected vector
    v_proj = P @ v
    v_proj_flat = v_proj.flatten()
    arrow_proj = Arrow3D(0, 0, 0, v_proj_flat[0], v_proj_flat[1], v_proj_flat[2],
                         mutation_scale=15, lw=2.5, arrowstyle="->", 
                         color=colors[i], linestyle='--', alpha=0.7)
    ax.add_artist(arrow_proj)
    
    # Connection line (showing perpendicular component)
    ax.plot([v_flat[0], v_proj_flat[0]], 
            [v_flat[1], v_proj_flat[1]], 
            [v_flat[2], v_proj_flat[2]], 
            'k--', alpha=0.3, linewidth=1)

# Set labels and title
ax.set_xlabel('$x_1$', fontsize=12, fontweight='bold')
ax.set_ylabel('$x_2$', fontsize=12, fontweight='bold')
ax.set_zlabel('$x_3$', fontsize=12, fontweight='bold')
ax.set_title('Orthogonal Projection onto Plane: $x_1 + 2x_2 - 2x_3 = 0$', 
             fontsize=14, fontweight='bold', pad=20)

# Add legend
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], color='lightblue', lw=6, label='Plane'),
    Line2D([0], [0], color='red', lw=3, label='Normal vector n'),
    Line2D([0], [0], color='gray', lw=2, label='Original vectors'),
    Line2D([0], [0], color='gray', lw=2, linestyle='--', label='Projected vectors'),
]
ax.legend(handles=legend_elements, loc='upper left', fontsize=11)

# Set viewing angle
ax.view_init(elev=20, azim=45)
ax.set_box_aspect([1,1,1])

plt.tight_layout()
plt.show()

print("\nInterpretation:")
print("• Blue surface: The plane x₁ + 2x₂ - 2x₃ = 0")
print("• Red arrow: Normal vector n = [1, 2, -2]")
print("• Solid colored arrows: Original vectors")
print("• Dashed colored arrows: Their projections onto the plane")
print("• Black dashed lines: The perpendicular component being removed")
```

## Geometric Intuition

### What is happening geometrically?

1. **The Plane**: All points $(x_1, x_2, x_3)$ satisfying $x_1 + 2x_2 - 2x_3 = 0$ form a 2D plane through the origin

2. **The Normal Vector**: $\mathbf{n} = [1, 2, -2]$ is perpendicular to this plane

3. **The Projection**: For any vector $\mathbf{v}$:
   - Decompose: $\mathbf{v} = \mathbf{v}_{\parallel} + \mathbf{v}_{\perp}$ (parallel + perpendicular to plane)
   - The perpendicular part is: $\mathbf{v}_{\perp} = \frac{\mathbf{n}^T \mathbf{v}}{\mathbf{n}^T \mathbf{n}} \mathbf{n}$
   - The projection is: $P\mathbf{v} = \mathbf{v}_{\parallel} = \mathbf{v} - \mathbf{v}_{\perp}$

Let's visualize this decomposition:

```python
# Choose a specific vector to decompose
v = np.array([[3], [1], [2]])

# Compute projections
v_parallel = P @ v  # Projection onto plane
v_perp = v - v_parallel  # Perpendicular component

# Alternative: direct computation of perpendicular component
v_perp_direct = (n @ n.T @ v) / (n.T @ n)[0, 0]

print("Vector Decomposition")
print("="*50)
print(f"Original vector v = {v.T}")
print(f"\nParallel component (on plane) = {v_parallel.T}")
print(f"Perpendicular component = {v_perp.T}")
print(f"\nVerification: parallel + perpendicular = {(v_parallel + v_perp).T}")
print(f"Matches original? {np.allclose(v, v_parallel + v_perp)}")

# Check that perpendicular is parallel to normal
perp_normalized = v_perp / np.linalg.norm(v_perp)
n_normalized = n / np.linalg.norm(n)
print(f"\nPerpendicular component is parallel to normal? {np.allclose(np.abs(perp_normalized), np.abs(n_normalized))}")
```

## Summary

### The Projection Matrix

For the plane $x_1 + 2x_2 - 2x_3 = 0$, the orthogonal projection matrix is:

$$P = \begin{bmatrix} 8/9 & -2/9 & 2/9 \\ -2/9 & 5/9 & 4/9 \\ 2/9 & 4/9 & 5/9 \end{bmatrix}$$

### Key Insights

1. **The matrix projects any vector onto the plane** by removing its component perpendicular to the plane

2. **The perpendicular component** is always parallel to the normal vector $\mathbf{n} = [1, 2, -2]$

3. **Properties verified**:
   - $P^2 = P$ (idempotent)
   - $P^T = P$ (symmetric)
   - $P\mathbf{n} = \mathbf{0}$ (normal is in the kernel)
   - For all $\mathbf{v}$: $P\mathbf{v}$ lies on the plane

4. **Geometric interpretation**: 
   - Every vector can be decomposed as $\mathbf{v} = \mathbf{v}_{\parallel} + \mathbf{v}_{\perp}$
   - The projection matrix extracts the parallel component: $P\mathbf{v} = \mathbf{v}_{\parallel}$
